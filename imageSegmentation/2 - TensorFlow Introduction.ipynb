# **Basic Tensor Operations (tf.Tensor)**

# **1**. **Constant: scalar, vector,matrix**

**tf.constant(value, dtype=None, shape=None, name='Const') **

**value:** the Python object / list / NumPy array / tensor-like thing you want to convert.

**dtype:** (optional) to explicitly set the data type. If not set, TensorFlow infers it.

**Shape:** (optional) you can force the constant to have a particular shape; scalar values can be broadcast to fill that shape.

**name:**(optional) a name for the operation / tensor.



````python
import tensorflow as tf

# Scalar, vector, matrix
a = tf.constant(3)
b = tf.constant([1, 2, 3])
c = tf.constant([[1, 2], [3, 4]])

print(a, b, c)
print(a.numpy())
````


**Operations- constant**



````python
x = tf.constant([10, 20, 30])
y = tf.constant([1, 2, 3])

print(x + y)
print(x * y)
print(tf.square(x))
print(tf.reduce_sum(x))
````

# **2. Variables (tf.Variable)**

````python
w = tf.Variable(5.0)
print(w)

w.assign(10.0)
print(w)

w.assign_add(2.5)
print(w)
````

**Automatic Differentiation (tf.GradientTape)**

````python
x = tf.Variable(3.0)

with tf.GradientTape() as tape:
    y = x**2 + 2*x + 1

dy_dx = tape.gradient(y, x)
print("dy/dx =", dy_dx.numpy())
````

**Plot a Tensor (Line Plot)**

What it does

```tf.linspace(-3.0, 3.0, 300)```

Start: –3.0

End: 3.0

Number of points: 300

It returns a 1-D tensor of shape (300,) where the values increase linearly from –3.0 to 3.0 inclusive.

````python
import tensorflow as tf
import matplotlib.pyplot as plt

# Create a simple tensor
x = tf.linspace(-3.0, 3.0, 300)
y = tf.math.sin(x)

# Plot
plt.plot(x.numpy(), y.numpy())
plt.title("Sine Curve")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True)
plt.show()
````

**Basic Histogram From a TensorFlow Vector**

````python
import tensorflow as tf
import matplotlib.pyplot as plt

# Example vector (1-D tensor)
vec = tf.random.normal((1000,))   # 1000 values

# Plot histogram
plt.hist(vec.numpy(), bins=30, color='skyblue', edgecolor='black')
plt.title("Histogram of TensorFlow Vector")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
````

**3D Line Plot**

````python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Create a 3D curve
t = tf.linspace(0.0, 10.0, 300)
x = tf.math.sin(t)
y = tf.math.cos(t)
z = t

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot(x.numpy(), y.numpy(), z.numpy())
ax.set_title("3D Line Plot: Helix")
plt.show()
````

**Voxel Plot (3D Volume Visualization)**

Useful for 3D medical images, 3D occupancy grids, etc.

ax.voxels draws a 3D arrangement of cubes based on a boolean grid.

````python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
import tensorflow as tf

# 3D tensor (10×10×10)
volume = tf.random.uniform((10,10,10)) > 0.7  # boolean voxels creates a 10×10×10 boolean tensor, where each element is: True if the random number > 0.7 False otherwise

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.voxels(volume.numpy(), edgecolor='k')
plt.title("3D Voxel Plot")
plt.show()
````
